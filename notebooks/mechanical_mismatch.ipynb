# Brain-electrode mechanical mismatch

Visualizes the elastic modulus of conventional electrode materials versus brain tissue, using the ranges reported in Kozai et al. (2015).

Ranges used: conventional materials (silicon, tungsten, platinum) 100-200 GPa; brain tissue 1-10 kPa.

```python
import numpy as np
import matplotlib.pyplot as plt

# Ranges in pascals (Pa)
materials = {
    "Conventional electrode\n(Si, W, Pt)": (100e9, 200e9),
    "Brain tissue": (1e3, 10e3),
}

fig, ax = plt.subplots(figsize=(8, 3.5))
for i, (name, (low, high)) in enumerate(materials.items()):
    ax.barh(i, high - low, left=low, height=0.5)
ax.set_xscale("log")
ax.set_yticks(range(len(materials)))
ax.set_yticklabels(materials.keys())
ax.set_xlabel("Elastic modulus (Pa, log scale)")
ax.set_title("Elastic modulus: electrode materials vs brain tissue")
plt.tight_layout()
plt.show()
```

## How many orders of magnitude?

```python
e_low, e_high = 100e9, 200e9   # Pa
t_low, t_high = 1e3, 10e3      # Pa

min_ratio = e_low / t_high
max_ratio = e_high / t_low
print(f"Stiffness ratio: {min_ratio:.0e} to {max_ratio:.0e}")
print(f"Orders of magnitude: {np.log10(min_ratio):.1f} to {np.log10(max_ratio):.1f}")
```
